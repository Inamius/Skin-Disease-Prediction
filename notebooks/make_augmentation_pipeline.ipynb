```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==========================
# CHANGE IMAGE PATH
# ==========================
IMAGE_PATH = "sample_lesion.jpg"

img = cv2.imread(IMAGE_PATH)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img = cv2.resize(img, (224,224))

# rescale visualization
rescaled = img.copy()

# rotate 15 deg
M = cv2.getRotationMatrix2D((112,112), 15, 1.0)
rotated = cv2.warpAffine(img, M, (224,224))

# zoom 10%
zoom = cv2.resize(img, None, fx=1.1, fy=1.1)
zoom = zoom[12:236, 12:236]
zoom = cv2.resize(zoom, (224,224))

# horizontal flip
flipped = cv2.flip(img, 1)

# ==========================
# PLOT
# ==========================
fig, ax = plt.subplots(2, 2, figsize=(10,10))

ax[0,0].imshow(img)
ax[0,0].set_title("Original Image", fontweight="bold")

ax[0,1].imshow(rotated)
ax[0,1].set_title("Rotation (15°)", fontweight="bold")

ax[1,0].imshow(zoom)
ax[1,0].set_title("Zoom (10%)", fontweight="bold")

ax[1,1].imshow(flipped)
ax[1,1].set_title("Horizontal Flip", fontweight="bold")

for row in ax:
    for a in row:
        a.axis("off")

plt.tight_layout()

plt.savefig(
    "Figure_3_3_Data_Augmentation.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: Figure_3_3_Data_Augmentation.png")
```
