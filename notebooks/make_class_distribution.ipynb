```python
import pandas as pd
import matplotlib.pyplot as plt

# ==========================
# CHANGE PATH IF NEEDED
# ==========================
CSV_PATH = "../data/HAM10000_metadata.csv"

# ==========================
# LOAD CSV
# ==========================
df = pd.read_csv(CSV_PATH)

print(df.head())

# ==========================
# CLASS MAPPING
# ==========================
label_map = {
    "nv": "Melanocytic nevi",
    "mel": "Melanoma",
    "bkl": "Benign keratosis",
    "bcc": "Basal cell carcinoma",
    "akiec": "Actinic keratoses",
    "vasc": "Vascular lesions",
    "df": "Dermatofibroma"
}

df["diagnosis"] = df["dx"].map(label_map)

counts = df["diagnosis"].value_counts()

print(counts)

# ==========================
# PLOT
# ==========================
plt.figure(figsize=(12,7))

bars = plt.bar(
    counts.index,
    counts.values
)

plt.title(
    "Class Distribution of HAM10000 Dataset",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel(
    "Diagnostic Class",
    fontsize=13
)

plt.ylabel(
    "Number of Images",
    fontsize=13
)

plt.xticks(
    rotation=25,
    ha="right"
)

# numbers on top
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height + 50,
        f"{int(height)}",
        ha="center",
        fontsize=11,
        fontweight="bold"
    )

plt.tight_layout()

plt.savefig(
    "Figure_3_1_Class_Distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: Figure_3_1_Class_Distribution.png")
```
